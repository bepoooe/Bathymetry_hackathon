```python
import pandas as pd

# Define the path to your CSV file (replace with your actual path)
csv_file_path = '/content/Glacier Lake Outburst Flood Database V4.0.csv'  # Change this to the actual path of your CSV file

# Read the CSV file using pandas
df = pd.read_csv(csv_file_path)

# Convert the DataFrame into a list of dictionaries
glacial_lakes = df.to_dict(orient='records')

# Format the output as a JavaScript array of objects
print("const mockGlacialLakes = [")
for i, lake in enumerate(glacial_lakes):
    print("    {")
    for key, value in lake.items():
        # Clean up the key and format values correctly
        key = key.strip()  # Remove extra spaces from the key if any
        if isinstance(value, str):
            value = f"'{value.strip()}'"  # Add quotes around string values
        print(f"      '{key}': {value},")
    print(f"    }}{',' if i < len(glacial_lakes) - 1 else ''}")
print("  ];")
```
